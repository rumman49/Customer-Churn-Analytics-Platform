###  Renaming the column names adding removing spaces and adding '_'

```python
df.columns = df.columns.str.lower().str.replace(' ', '_')
df.columns = df.columns.str.lower().str.replace('-', '_')
df.columns
```

### Change Datatypes


```python
date_columns = ['order_date', 'ship_date',]

for col in date_columns:
    df[col] = pd.to_datetime(df[col])
```

```python
df.info()
```

### Blank/Whitespace Strings in Object Columns

```python
text_cols = df.select_dtypes(include=['object']).columns
for col in text_cols:
    empty_strings = df[col].astype(str).str.strip().eq('').sum()
    if empty_strings > 0:
        print(f"Column '{col}' has {empty_strings} empty/whitespace string entries.")
        
```

```python
df.loc[df['sales']<0] 
df.loc[df['quantity']<0] 
df.loc[df['discount']<0] 
df.loc[df['profit']<0].head(5) # found negative numbers in profit 
```

### Saving the cleaned file

```python
cols = ['row_id','order_id', 'order_date', 'ship_date', 'ship_mode', 'customer_id',
       'customer_name', 'segment', 'country', 'city', 'state', 'postal_code',
       'region', 'product_id', 'category', 'sub_category', 'product_name',
       'sales', 'quantity', 'discount', 'profit', 'cost', 'profit_margin',
       'shipping_days', 'order_year', 'order_month', 'order_quarter',
       'order_days']

for col in cols:
    max_char_len = df[col].astype(str).str.len().max()
    print(f"{col}: {max_char_len}")
```

```python
df.to_csv('D:\Customer-Churn-Analytics\Data\cleaned_superstore.csv', index=False)
```

### Feature Engineering 

```python
# cost column
df['cost'] = (df['sales'] - df['profit'])

```

```python
# profit_margin column
df['profit_margin'] = (df['profit'] / df['sales'] * 100)
```

```python
df['shipping_days'] = (df['ship_date'] - df['order_date']).dt.days
```

```python
df['order_year'] = df['order_date'].dt.year
df['order_month'] = df['order_date'].dt.month
df['order_quarter'] = df['order_date'].dt.quarter
df['order_days'] = df['order_date'].dt.day
```
